## Data pre-processing

Manage null values

```python
df.columns
```

```python
 #df.drop('BMI', axis=1, inplace=True)
df = df.drop('BMI', axis=1)
```

```python
# df.drop('smoking', axis=1, inplace=True)
df = df.drop('Smoking', axis=1)
```

```python
df.isnull().sum()
```

```python
def fillnaIntMode(cols):
    for i in cols:
        df[i] = df[i].fillna(df[i].mode()[0])

columns = ['Sex','Age','Level_of_Stress','Pregnancy','Physical_activity','Chronic_kidney_disease']
fillnaIntMode(columns)
```

```python
df.info()
```

##Manage data type

```python
df['Sex'] = df['Sex'].astype('int64')
# df['price'].astype('int64', inplace=True)
```

```python
df.info()
```

```python
 !pip install scikit-learn

 import numpy as np
 from sklearn.preprocessing import LabelEncoder
```

```python
 data = pd.DataFrame({'Sex':df['Sex'].unique()})
 data_LE = LabelEncoder()
 data_LE.fit(np.ravel(data))    # data_LE.fit(data)
 df['Sex'] = data_LE.transform(df['Sex'])
```

```python
def convertObjToInt(cols):
    for i in cols:
        data = pd.DataFrame({i:df[i].unique()})
        data_LE = LabelEncoder()
        data_LE.fit(np.ravel(data))    # data_LE.fit(data)
        df[i] = data_LE.transform(df[i])
        
columns = ['Sex', 'Age', 'Pregnancy']
convertObjToInt(columns)
```

```python
import pandas as pd
df = pd.read_csv('cleaned_data.csv')
# df.info()
df.columns
```

```python
X = df.drop('Age', axis=1)   # Dependant columns / Input columns
y = df['Age']                # Resultant columns / Output column
```

```python
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(X,y, test_size=0.3, shuffle=False) # 80:20, 70:30, 60:40
```

```python
print(train_X.shape, train_y.shape)
print(test_X.shape, test_y.shape)
```

##Applying Classifier

```python
from sklearn.svm import SVC # Support Vector Classifier

model_svc = SVC()
model_svc.fit(train_X, train_y)

print(model_svc)
```

```python
model_pred_svc = model_svc.predict(test_X)
model_pred_svc
```

```python
from sklearn.metrics import accuracy_score
model_acc_svc = accuracy_score(test_y, model_pred_svc)
print(round(model_acc_svc,3)*100, "%")
```
